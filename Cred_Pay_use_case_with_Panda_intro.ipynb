# Problem Definition

## Business Context
Cred-Pay is a consultation firm that partners with banks and checks if their customers are eligible for a credit card. They are a startup just at the roots of starting their business. They have partnered with a few banks and are currently collecting data for credit card applications.

## Objective
You have been hired as a Data Scientist to handle and organize the data. You are responsible to organise the data so that, it will be easily accessible and help the company predict, if the application can be accepted for a credit card or not.

## Solution Approach
You have tried to understand how the credit card works and what all factors come into play. You can organize the data so that the company can help banks to provide credit cards to the customers. You have considered some features related to credit card applications. Some of them are Customer ID, Credit score, age of customer, debt status, and the number of existing credit cards.



```python
cust_id = ['CUST_001','CUST_002','CUST_003','CUST_004','CUST_005']
credit_score = [620, 750, 780, 500, 655]
age = [25, 30, 35, 40, 45]
debt_status = ['no-debt', 'in-debt', 'no-debt', 'in-debt', 'no-debt']
credit_cards_count = [2, 3, 4, 1, 2]
```

```python
print('Customer Ids:', cust_id)
print('Credit Score:', credit_score)
print('Customer Age:', age)
print('Debt Status:', debt_status)
print('No of Credit Cards:', credit_cards_count)
```

Now, If we want to get all the details of the customer whose id is "CUST_002"

```python
position = cust_id.index('CUST_002')

print('Customer Id:', cust_id[position])
print('Credit Score:', credit_score[position])
print('Customer Age:', age[position])
print('Debt Status:', debt_status[position])
print('No of Credit Cards:', credit_cards_count[position])
```

We just received information that the debt status for **CUST_001** needs to be changed as the customer is in debt now. Let's update the same in the list.

```python
position = cust_id.index('CUST_001')
debt_status[position] = 'in-debt'

print('Debt Status:', debt_status)
```

We recently got an update from Cred-Pay that we need to update the credit score of all users. The Credit score is calculated based on 4 factors:

Payment History(High impact)
Credit Utilisation Ratio(High impact)
Age of the Credit(Medium impact)
Total Accounts(Low impact)

Since Payment History has a huge impact on the Credit Score we have to update the score based on the user's history of on-time payment.

For this, first, we have to create a list called prec_on_time, which stores the percentage of on-time payments done by the user. If a user's prec_on_time is greater than 0.7, then we have to increase their credit score by 2%.

Let's also print the customerid for those, whose credit score will be updated. Let's try this out.

```python
prec_on_time = [0.62, 0.78, 0.20, 0.32, 0.85]
```

```python
for i in range(0, len(prec_on_time)):
    if prec_on_time[i] > 0.7:
        position = i
        print('Customer Id:', cust_id[position])
        credit_score
```

Great! Now, we will be focusing on our real assignemnt to increase the elligible customer credit score by 2%, else print "They are not eligible for credit score upgrade."

```python
for i in range(0, len(prec_on_time)):
    if prec_on_time[i] > 0.7:
        credit_score[i] = round(credit_score[i] * 1.02)
        print('Customer Id:', cust_id[i])
        print('Credit Score:', credit_score[i])
    else:
        print('They are not eligible for credit score upgrade.')
```

Let's find out how many people have credit score greater than 700?

```python
count = 0
for i in range(0, len(credit_score)):
    if credit_score[i] > 700:
        count += 1

count
```

Can we check How many people have more than 2 credit cards?

```python
count = 0

for i in range(0, len(credit_cards_count)):
    if credit_cards_count[i] > 2:
        count += 1

count
```

Can we count how many people are in debt and how many are in not debt ?

```python
in_debt = 0
not_in_debt = 0

for i in range(0, len(debt_status)):
    if debt_status[i] != 'in-debt':
        not_in_debt += 1
    else:
        in_debt += 1

print('In Debt:', in_debt)
print('Not In Debt:', not_in_debt)

count
```

Let's try to print all the information of customer whose customer id = 'CUST_002'

```python
for i in range(0, len(cust_id)):
    if cust_id[i] == 'CUST_002':
        print('Customer Id:', cust_id[i])
        print('Credit Score:', credit_score[i])
        print('Customer Age:', age[i])
        print('Debt Status:', debt_status[i])
        print('No of Credit Cards:', credit_cards_count[i])
```

Cred-Pay has informed us that if a customer's percentage of on-time payments (prec_on_time) is less than 40% and the customer has more than 2 credit cards (no_of_existing_credit_cards), then their credit score should be decreased by 5%.

```python
for i in range(0, len(prec_on_time)):
    if prec_on_time[i] < 0.4 and credit_cards_count[i] > 2:
        credit_score[i] = round(credit_score[i] * 0.95)
        print('Customer Id:', cust_id[i])
        print('Credit Score:', credit_score[i])

print(credit_score)
```

Create a function to display customer info as per id

```python
def customer_info(customer_id):
    pos = cust_id.index(customer_id)
    print('Customer Id:', cust_id[pos])
    print('Credit Score:', credit_score[pos])
    print('Customer Age:', age[pos])
    print('Debt Status:', debt_status[pos])
    print('No of Credit Cards:', credit_cards_count[pos])
```

```python
customer_info('CUST_002')
```

Now, lets use dictionary instead of having different list, and create functions to resolve upcoming queries

```python
cred_user_dict = {
  'cust_id': cust_id,
  'credit_score': credit_score,
  'age': age,
  'debt_status': debt_status,
  'credit_cards_count': credit_cards_count,
  'prec_on_time': prec_on_time
}
cred_user_dict
```

**Let's get all the information for the Customer whose id is 'cid-0004'**

```python
def get_customer_info(cid):
    pos = cred_user_dict['cust_id'].index(cid)
    for key, value in cred_user_dict.items():  # Iterating through all the keys of dictionary, with values index as constant
      print(key ,value[pos], sep=': ')

get_customer_info('CUST_004')
```

Lets get one step ahead and instead of creating key of each attribute, use dict key as customer_id and value will be related attribute values.

```python
cred_str_user_dict = {}

for i in range(len(cust_id)):
  cred_str_user_dict[cust_id[i]] = [credit_score[i], age[i], debt_status[i], credit_cards_count[i], prec_on_time[i]]

print('Old structure: ', cred_user_dict)
print('New structure: ', cred_str_user_dict)
```
