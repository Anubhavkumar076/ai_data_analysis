# Genres that are most liked by the users

```python
df_merge = movie.merge(ratings, on = "movie id", how= "inner")
```

```python
df_merge.head()
```

```python
df_merge[df_merge["rating"]>3]
```

```python
for genre in genres:
  print(genre, df_merge[df_merge[genre]==1].rating.mean())
```

# Movie with their average rating

```python
avg_rating = df_merge.groupby(['movie title'])['rating'].mean().reset_index() # reset_index will reset the index of dataframe to default indexing (0 to number of rows minus 1)
avg_rating
```

```python
avg_rating.rename(columns={'rating':'Avg_rating'},inplace=True)
avg_rating.head()
```

```python
avg_rating.sort_values('Avg_rating',ascending=False)
```

```python
avg_rating[avg_rating["Avg_rating"]==5]
```

# which movies are rated most of the times?

```python
movie_count = df_merge.groupby('movie title')['rating'].count().reset_index()
movie_count
```

```python
movie_count.sort_values('rating',ascending=False, inplace=True)
movie_count
```

```python
movie_count.rename(columns={'rating': 'Rating Counts'}, inplace=True)
movie_count
```

```python
movie_count100 = movie_count[movie_count['Rating Counts'] > 100]
movie_count100
```

```python
avg_rating.head()
```

```python
movie_count100.head()
```

```python
top_movie_by_avg = avg_rating.merge(movie_count100, on='movie title', how='inner')
top_movie_by_avg
```

```python
top_movie_by_avg.sort_values('Avg_rating', ascending=False)
```
